3. Analizar qué colonias tienen el promedio de alquiler más alto.


```python
datos.groupby('Colonia')[['Valor']].mean().sort_values('Valor',ascending=False)
```

```python
# Extracción de datos para validar resultado general
datos[datos['Colonia'] == 'Las Aguilas']
```

```python
datos[datos['Colonia'] == 'Las Aguilas'].describe()
```

4. Crear un gráfico de barras horizontales que muestre las 5 colonias con los promedios de alquiler más altos.

```python
# almaceno en variable
df_colonia_precio_promedio = datos.groupby('Colonia')[['Valor']].mean().sort_values('Valor',ascending=False).head(5)
# graficarlo con Pandas  - kind barh =barras horizontales, tamaño 12x8, color purpura
df_colonia_precio_promedio.plot(kind='barh',color='pink', title = 'Las 5 Colonias con los promedios de alquiler mas altos' )
```
